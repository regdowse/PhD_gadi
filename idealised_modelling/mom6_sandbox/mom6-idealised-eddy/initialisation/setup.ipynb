## Configuration and verified model conventions

The eddy is centred at the physical domain midpoint (250,000 m, 250,000 m), on the current 100 × 100, 5,000 m Cartesian grid. Its maximum-speed diameter is 100,000 m, not a cutoff diameter. The Gaussian has small nonzero tails at the closed walls; normal boundary flow is explicitly zeroed.

The saved `ocean_static.nc` supplies actual C-grid tracer/face coordinates, bathymetry, Coriolis and wet masks. `Vertical_coordinate.nc` supplies the 101 **nonuniform** reference interfaces. The generator checks these against current settings/topography; it does not reuse the old degree-grid horizontal coordinates.

For this source version (`build_versions.json`: MOM6 `776b61a60bff3ba9a44328402cb50a5524c3fbbe`), `ROTATION="betaplane"`, `F_0=-1e-4 s⁻¹` and the recorded default `BETA=0` give constant f. Thus cyclonic rotation is clockwise and ψ₀ is positive. If beta changes, centre-f balance is only an approximation and f variation and the resulting residual are reported.

The EOS uses **dimensional** `DRHO_DT=-0.2 kg m⁻³ K⁻¹`, `DRHO_DS=0.8 kg m⁻³ ppt⁻¹`, `DRHO_DP=0`, `RHO_T0_S0=1000 kg m⁻³`; the model Boussinesq reference density is **1035 kg m⁻³**, and g=9.8 m s⁻². Therefore α=−DRHO_DT/ρ₀, rather than assuming DRHO_DT itself is α.

The original `TS_CONFIG="linear"` varies T by **layer index**, not physical depth. The background here matches its 10-to-5°C nominal-layer profile, interpolated by physical depth below mean sea level (endpoint values held outside the nominal centres), with constant salinity 35 ppt. This avoids imposing deep-index temperatures in collapsed shallow-bottom layers.

## Balance and geometry

With depth $d$ measured downward from the local free surface,

$$\psi=\psi_0\max(1-d/H,0)\exp(-r^2/R_c^2),\quad H=1000\,\mathrm{m},\quad R_c=\sqrt{2}\,50000\,\mathrm{m}.$$

The convention is $u=-\psi_y$, $v=\psi_x$; $\operatorname{sign}(\psi_0)=-\operatorname{sign}(f)$ and $|\psi_0|\simeq16487.21\,\mathrm{m^2/s}$. At constant f, $\Phi'=f\psi$, $\mathrm{SSH}=f\psi(d=0)/g$ and $T'=f\,\psi_z/(g\alpha)$.

The generator evaluates velocities at **staggered face-layer physical depths**. It uses a layer-average $\psi_z$, including a partially intersected cutoff layer, so the hydrostatic column integral is consistent with the specified linear-decay function. Clipped reference interfaces reserve small positive collapsed-bottom layers; z-star dilation adds SSH **once** while fixing the local bottom. `eta[0]=ssh`, `eta[-1]=-depth`.

This is **approximate geostrophic initialization**, not a steady nonlinear vortex: centrifugal acceleration (Rossby number about 0.04), finite-grid pressure gradients, closed-wall tails, free-surface slope corrections and any beta variation remain. The reported geostrophic residual uses centred interior stencils at fixed physical z, away from the cutoff and shelves; it is not the compiled MOM6 pressure-gradient operator. The hydrostatic residual checks the eddy anomaly column integral, not a complete discrete solver equilibrium.
